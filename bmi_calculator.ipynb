# BMI Calculator

Calculate Body Mass Index (BMI) using weight in pounds and height in inches. This notebook applies the same category and health-risk ranges as the command-line and browser calculators.

Formula: `BMI = (weight in pounds × 703) / (height in inches)²`

```python
import math

name = input("Enter your name: ").strip() or "Friend"

try:
    weight = float(input("Enter your weight in pounds: "))
    height = float(input("Enter your height in inches: "))
except ValueError:
    print("Error: Please enter valid numeric values for weight and height.")
else:
    if not math.isfinite(weight) or not math.isfinite(height) or weight <= 0 or height <= 0:
        print("Error: Please enter valid positive inputs.")
    else:
        bmi = (weight * 703) / (height ** 2)
        categories = [
            (18.5, "Underweight", "Minimal"),
            (25, "Normal weight", "Minimal"),
            (30, "Overweight", "Increased"),
            (35, "Obese", "High"),
            (40, "Severely obese", "Very high"),
            (float("inf"), "Morbidly obese", "Extremely high"),
        ]
        category, risk = next((category, risk) for upper_bound, category, risk in categories if bmi < upper_bound)
        print(f"\n{name}, your BMI is: {bmi:.2f}")
        print(f"Category: {category}")
        print(f"Health risk: {risk}")
        print("\n*Disclaimer: This is a general screening number, not medical advice.*")
```

## BMI categories

| BMI | Category | Health risk |
| --- | --- | --- |
| Below 18.5 | Underweight | Minimal |
| 18.5 to below 25 | Normal weight | Minimal |
| 25 to below 30 | Overweight | Increased |
| 30 to below 35 | Obese | High |
| 35 to below 40 | Severely obese | Very high |
| 40 and above | Morbidly obese | Extremely high |

BMI is a general screening measure, not a diagnosis or a substitute for medical advice.
